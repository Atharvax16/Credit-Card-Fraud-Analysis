## Improved Models — Using All Features, Scaling, and Tuned Hyperparameters

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_recall_curve
from lightgbm import LGBMClassifier
from imblearn.over_sampling import SMOTE

# Use ALL features (drop only Class and Hour)
X_all = data.drop(columns=['Class', 'Hour'])
y_all = data['Class']

X_train_all, X_test_all, y_train_all, y_test_all = train_test_split(
    X_all, y_all, test_size=0.2, random_state=42, stratify=y_all
)

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_all)
X_test_scaled = scaler.transform(X_test_all)

# SMOTE on scaled training data
smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train_scaled, y_train_all)

print(f"Resampled training set: {pd.Series(y_train_res).value_counts().to_dict()}")

# Tuned models
improved_models = {
    'Logistic Regression': LogisticRegression(
        max_iter=1000, random_state=42, class_weight='balanced', C=0.1
    ),
    'Random Forest (Tuned)': RandomForestClassifier(
        n_estimators=300, max_depth=20, min_samples_split=5,
        class_weight='balanced_subsample', random_state=42, n_jobs=-1
    ),
    'LightGBM (Tuned)': LGBMClassifier(
        n_estimators=300, max_depth=8, learning_rate=0.05,
        scale_pos_weight=len(y_train_all[y_train_all==0]) / max(len(y_train_all[y_train_all==1]), 1),
        reg_alpha=0.1, reg_lambda=1.0, random_state=42, verbose=-1
    ),
}

improved_results = []

for name, mdl in improved_models.items():
    mdl.fit(X_train_res, y_train_res)
    y_prob_imp = mdl.predict_proba(X_test_scaled)[:, 1]

    # Find optimal threshold that maximizes F1
    prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, y_prob_imp)
    f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
    best_idx = np.argmax(f1_arr)
    best_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

    y_pred_imp = (y_prob_imp >= best_thresh).astype(int)

    report = classification_report(y_test_all, y_pred_imp, output_dict=True)
    recall = report['1']['recall']
    precision = report['1']['precision']
    f1 = report['1']['f1-score']
    auc = roc_auc_score(y_test_all, y_prob_imp)

    improved_results.append({
        'Model': name,
        'Threshold': round(best_thresh, 3),
        'Precision': round(precision, 4),
        'Recall': round(recall, 4),
        'F1 Score': round(f1, 4),
        'ROC AUC': round(auc, 4)
    })

    print(f"\n{'='*50}")
    print(f"{name} (optimal threshold={best_thresh:.3f}):")
    print(confusion_matrix(y_test_all, y_pred_imp))
    print(classification_report(y_test_all, y_pred_imp))
    print(f"ROC AUC: {auc:.4f}")

improved_df = pd.DataFrame(improved_results).sort_values('F1 Score', ascending=False)
print("\n" + "="*60)
print("IMPROVED MODEL COMPARISON (with threshold tuning):")
print("="*60)
print(improved_df.to_string(index=False))
```

## Isolation Forest — Unsupervised Anomaly Detection

```python
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_recall_curve

# Isolation Forest works on the ORIGINAL data (no SMOTE needed — it's unsupervised)
# Using the same scaled train/test split from the improved section above

iso_forest = IsolationForest(
    n_estimators=300,
    max_samples='auto',
    contamination=0.00173,  # matches the actual fraud ratio in dataset (~0.173%)
    random_state=42,
    n_jobs=-1
)

# Fit on training data only
iso_forest.fit(X_train_scaled)

# Predict: Isolation Forest returns -1 for anomalies, 1 for normal
iso_preds_raw = iso_forest.predict(X_test_scaled)
# Convert: -1 (anomaly) → 1 (fraud), 1 (normal) → 0 (legit)
iso_preds = np.where(iso_preds_raw == -1, 1, 0)

# Anomaly scores (more negative = more anomalous)
iso_scores = -iso_forest.decision_function(X_test_scaled)  # negate so higher = more fraudulent

print("Isolation Forest Results:")
print("="*50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, iso_preds))
print("\nClassification Report:")
print(classification_report(y_test_all, iso_preds))
print(f"ROC AUC Score: {roc_auc_score(y_test_all, iso_scores):.4f}")

# Threshold tuning on anomaly scores for better F1
prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, iso_scores)
f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
best_idx = np.argmax(f1_arr)
best_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.0

iso_preds_tuned = (iso_scores >= best_thresh).astype(int)
print(f"\nAfter Threshold Tuning (threshold={best_thresh:.4f}):")
print("="*50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, iso_preds_tuned))
print("\nClassification Report:")
print(classification_report(y_test_all, iso_preds_tuned))

report = classification_report(y_test_all, iso_preds_tuned, output_dict=True)
print(f"Recall: {report['1']['recall']:.4f}  |  F1: {report['1']['f1-score']:.4f}  |  Precision: {report['1']['precision']:.4f}")
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_recall_curve

X_train_tensor = torch.FloatTensor(X_train_res)
y_train_tensor = torch.FloatTensor(y_train_res.values if hasattr(y_train_res, 'values') else y_train_res)
X_test_tensor = torch.FloatTensor(X_test_scaled)

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_loader = DataLoader(train_dataset, batch_size=256, shuffle=True)

class FraudMLP(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.3),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.BatchNorm1d(64),
            nn.Dropout(0.3),

            nn.Linear(64, 32),
            nn.ReLU(),
            nn.BatchNorm1d(32),
            nn.Dropout(0.2),

            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x).squeeze()

model_mlp = FraudMLP(X_train_scaled.shape[1])
print(model_mlp)

# Class weight for loss function
n_neg = np.sum(y_train_res == 0)
n_pos = np.sum(y_train_res == 1)
pos_weight = torch.FloatTensor([n_neg / n_pos])
criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
optimizer = torch.optim.Adam(model_mlp.parameters(), lr=0.001)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=5, factor=0.5)

# Training loop with early stopping
best_loss = float('inf')
patience, patience_counter = 10, 0
train_losses = []
val_losses = []

# Split a validation set from training data
val_size = int(0.15 * len(X_train_tensor))
indices = torch.randperm(len(X_train_tensor))
val_idx, train_idx = indices[:val_size], indices[val_size:]

X_val_t = X_train_tensor[val_idx]
y_val_t = y_train_tensor[val_idx]
train_sub = TensorDataset(X_train_tensor[train_idx], y_train_tensor[train_idx])
train_loader = DataLoader(train_sub, batch_size=256, shuffle=True)

for epoch in range(100):
    model_mlp.train()
    epoch_loss = 0
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        output = model_mlp(X_batch)
        loss = criterion(output, y_batch)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_train_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_train_loss)

    # Validation
    model_mlp.eval()
    with torch.no_grad():
        val_output = model_mlp(X_val_t)
        val_loss = criterion(val_output, y_val_t).item()
    val_losses.append(val_loss)
    scheduler.step(val_loss)

    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}/100 — Train Loss: {avg_train_loss:.4f} | Val Loss: {val_loss:.4f}")

    # Early stopping
    if val_loss < best_loss:
        best_loss = val_loss
        patience_counter = 0
        best_state = model_mlp.state_dict().copy()
    else:
        patience_counter += 1
        if patience_counter >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break

model_mlp.load_state_dict(best_state)
print("Training complete!")
```

```python
# 1. OPTUNA HYPERPARAMETER TUNING FOR LIGHTGBM
# Optuna automatically searches for the best hyperparameters that maximize fraud detection F1-score.
# This replaces manual tuning and can significantly boost model performance.

import optuna
from lightgbm import LGBMClassifier
from sklearn.metrics import f1_score, precision_recall_curve, roc_auc_score
from sklearn.metrics import classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')
optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 500),
        'max_depth': trial.suggest_int('max_depth', 3, 15),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 50),
        'scale_pos_weight': trial.suggest_float('scale_pos_weight', 1.0, 500.0, log=True),
        'random_state': 42,
        'verbose': -1
    }
    
    model = LGBMClassifier(**params)
    model.fit(X_train_res, y_train_res)
    y_prob = model.predict_proba(X_test_scaled)[:, 1]
    
    # Optimize for best F1 via threshold tuning
    prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, y_prob)
    f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
    best_f1 = np.max(f1_arr)
    
    return best_f1

# Run optimization (50 trials)
study = optuna.create_study(direction='maximize', study_name='lgbm_fraud')
study.optimize(objective, n_trials=50, show_progress_bar=True)

print(f"Best F1 Score: {study.best_value:.4f}")
print(f"\nBest Hyperparameters:")
for key, val in study.best_params.items():
    print(f"  {key}: {val}")
```

```python
# Train the Optuna-optimized LightGBM model
best_params = study.best_params
best_params['random_state'] = 42
best_params['verbose'] = -1

lgbm_optuna = LGBMClassifier(**best_params)
lgbm_optuna.fit(X_train_res, y_train_res)

y_prob_optuna = lgbm_optuna.predict_proba(X_test_scaled)[:, 1]

# Find optimal threshold
prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, y_prob_optuna)
f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
best_idx = np.argmax(f1_arr)
optuna_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

y_pred_optuna = (y_prob_optuna >= optuna_thresh).astype(int)

print(f"Optuna-Tuned LightGBM (threshold={optuna_thresh:.4f}):")
print("=" * 50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, y_pred_optuna))
print("\nClassification Report:")
print(classification_report(y_test_all, y_pred_optuna))
print(f"ROC AUC: {roc_auc_score(y_test_all, y_prob_optuna):.4f}")
```

```python
# 2. COST-SENSITIVE LEARNING
# In fraud detection, missing a fraud (false negative) costs far more than a false alarm.
# We assign asymmetric misclassification costs to reflect real-world bank losses.

from sklearn.linear_model import LogisticRegression

# Missing fraud costs 100x more than a false alarm
cost_weights = {0: 1, 1: 100}

model_cost = LogisticRegression(max_iter=1000, random_state=42, class_weight=cost_weights)
model_cost.fit(X_train_res, y_train_res)

y_prob_cost = model_cost.predict_proba(X_test_scaled)[:, 1]

# Threshold tuning for cost-sensitive model
prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, y_prob_cost)
f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
best_idx = np.argmax(f1_arr)
cost_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

y_pred_cost = (y_prob_cost >= cost_thresh).astype(int)

print("Cost-Sensitive Logistic Regression:")
print(f"Misclassification cost ratio - FN:FP = 100:1")
print(f"Optimal threshold: {cost_thresh:.4f}")
print("=" * 50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, y_pred_cost))
print("\nClassification Report:")
print(classification_report(y_test_all, y_pred_cost))
print(f"ROC AUC: {roc_auc_score(y_test_all, y_prob_cost):.4f}")
```

```python
# 3. MODEL CALIBRATION
# Raw model probabilities may not reflect true fraud likelihood.
# Calibration ensures that when the model says "80% fraud", it is actually ~80%.
# This is critical for banks making risk-based decisions.

from sklearn.calibration import CalibratedClassifierCV, calibration_curve

# Calibrate the Optuna-tuned LightGBM using isotonic regression
calibrated_lgbm = CalibratedClassifierCV(lgbm_optuna, method='isotonic', cv=5)
calibrated_lgbm.fit(X_train_res, y_train_res)

y_prob_calibrated = calibrated_lgbm.predict_proba(X_test_scaled)[:, 1]

# Compare calibration before vs after
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for name, probs, ax_idx in [("Before Calibration", y_prob_optuna, 0),
                              ("After Calibration", y_prob_calibrated, 1)]:
    fraction_pos, mean_predicted = calibration_curve(y_test_all, probs, n_bins=10)
    axes[ax_idx].plot(mean_predicted, fraction_pos, 's-', label='Model')
    axes[ax_idx].plot([0, 1], [0, 1], 'k--', label='Perfect')
    axes[ax_idx].set_xlabel('Mean Predicted Probability')
    axes[ax_idx].set_ylabel('Fraction of Positives')
    axes[ax_idx].set_title(name)
    axes[ax_idx].legend()
    axes[ax_idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Evaluate calibrated model with threshold tuning
prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, y_prob_calibrated)
f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
best_idx = np.argmax(f1_arr)
cal_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

y_pred_cal = (y_prob_calibrated >= cal_thresh).astype(int)

print(f"\nCalibrated LightGBM (threshold={cal_thresh:.4f}):")
print("=" * 50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, y_pred_cal))
print("\nClassification Report:")
print(classification_report(y_test_all, y_pred_cal))

auc_before = roc_auc_score(y_test_all, y_prob_optuna)
auc_after = roc_auc_score(y_test_all, y_prob_calibrated)
print(f"ROC AUC Before Calibration: {auc_before:.4f}")
print(f"ROC AUC After Calibration:  {auc_after:.4f}")
```

### SHAP Interpretability
SHAP (SHapley Additive exPlanations) explains **why** the model flags a transaction as fraud. This is essential for regulatory compliance and for banks to justify fraud decisions to auditors and customers.

```python
# 4. SHAP INTERPRETABILITY
# SHAP values show which features push a prediction toward fraud vs legitimate.
# This gives transparency into the model decision-making process.

import shap

# Use a sample for SHAP (computationally expensive)
X_test_sample = pd.DataFrame(X_test_scaled, columns=X_all.columns).sample(200, random_state=42)

# SHAP explainer for the Optuna-tuned LightGBM
explainer = shap.TreeExplainer(lgbm_optuna)
shap_values = explainer.shap_values(X_test_sample)

# For binary classification, shap_values may be a list [class_0, class_1]
if isinstance(shap_values, list):
    shap_vals = shap_values[1]  # fraud class
else:
    shap_vals = shap_values

# Summary plot: which features matter most across all predictions
plt.figure(figsize=(10, 8))
shap.summary_plot(shap_vals, X_test_sample, show=False)
plt.title('SHAP Feature Importance - Fraud Detection (Optuna-Tuned LightGBM)')
plt.tight_layout()
plt.show()
```

```python
# SHAP Waterfall Plot - Explain a Single Fraud Prediction
# This shows exactly which features contributed to flagging one specific transaction.

fraud_mask = y_test_all.values == 1
fraud_indices_arr = np.where(fraud_mask)[0]

if len(fraud_indices_arr) > 0:
    fraud_idx = fraud_indices_arr[0]
    X_fraud_case = pd.DataFrame(X_test_scaled, columns=X_all.columns).iloc[[fraud_idx]]
    
    shap_case = explainer.shap_values(X_fraud_case)
    if isinstance(shap_case, list):
        shap_case_vals = shap_case[1]
    else:
        shap_case_vals = shap_case
    
    feature_contributions = pd.Series(shap_case_vals[0], index=X_all.columns)
    feature_contributions = feature_contributions.reindex(
        feature_contributions.abs().sort_values(ascending=True).index
    )
    
    colors = ['red' if x > 0 else 'blue' for x in feature_contributions]
    
    plt.figure(figsize=(10, 8))
    plt.barh(feature_contributions.index, feature_contributions.values, color=colors)
    plt.xlabel('SHAP Value (Red = pushes toward Fraud, Blue = pushes toward Legit)')
    plt.title('Feature Contributions for a Single Fraud Transaction')
    plt.tight_layout()
    plt.show()
    
    print("\nTop 5 features pushing this transaction toward FRAUD:")
    top_fraud = feature_contributions.sort_values(ascending=False).head(5)
    for feat, val in top_fraud.items():
        print(f"  {feat}: {val:+.4f}")
else:
    print("No fraud cases found in test set.")
```

```python
# SHAP Global Feature Importance - Mean absolute SHAP values
# Shows which features are most important OVERALL for fraud detection.

mean_shap = np.abs(shap_vals).mean(axis=0)
importance_df = pd.DataFrame({
    'Feature': X_all.columns,
    'Mean |SHAP|': mean_shap
}).sort_values('Mean |SHAP|', ascending=False)

plt.figure(figsize=(10, 8))
plt.barh(importance_df['Feature'][:15], importance_df['Mean |SHAP|'][:15], color='coral')
plt.xlabel('Mean |SHAP Value|')
plt.title('Top 15 Features by SHAP Importance')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

print("\nTop 10 Most Important Features for Fraud Detection:")
print(importance_df.head(10).to_string(index=False))
```
