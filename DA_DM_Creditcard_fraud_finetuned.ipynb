## Improved Models — Using All Features, Scaling, and Tuned Hyperparameters

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_recall_curve
from lightgbm import LGBMClassifier
from imblearn.over_sampling import SMOTE

# Use ALL features (drop only Class and Hour)
X_all = data.drop(columns=['Class', 'Hour'])
y_all = data['Class']

X_train_all, X_test_all, y_train_all, y_test_all = train_test_split(
    X_all, y_all, test_size=0.2, random_state=42, stratify=y_all
)

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_all)
X_test_scaled = scaler.transform(X_test_all)

# SMOTE on scaled training data
smote = SMOTE(sampling_strategy=0.5, random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train_scaled, y_train_all)

print(f"Resampled training set: {pd.Series(y_train_res).value_counts().to_dict()}")

# Tuned models
improved_models = {
    'Logistic Regression': LogisticRegression(
        max_iter=1000, random_state=42, class_weight='balanced', C=0.1
    ),
    'Random Forest (Tuned)': RandomForestClassifier(
        n_estimators=300, max_depth=20, min_samples_split=5,
        class_weight='balanced_subsample', random_state=42, n_jobs=-1
    ),
    'LightGBM (Tuned)': LGBMClassifier(
        n_estimators=300, max_depth=8, learning_rate=0.05,
        scale_pos_weight=len(y_train_all[y_train_all==0]) / max(len(y_train_all[y_train_all==1]), 1),
        reg_alpha=0.1, reg_lambda=1.0, random_state=42, verbose=-1
    ),
}

improved_results = []

for name, mdl in improved_models.items():
    mdl.fit(X_train_res, y_train_res)
    y_prob_imp = mdl.predict_proba(X_test_scaled)[:, 1]

    # Find optimal threshold that maximizes F1
    prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, y_prob_imp)
    f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
    best_idx = np.argmax(f1_arr)
    best_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

    y_pred_imp = (y_prob_imp >= best_thresh).astype(int)

    report = classification_report(y_test_all, y_pred_imp, output_dict=True)
    recall = report['1']['recall']
    precision = report['1']['precision']
    f1 = report['1']['f1-score']
    auc = roc_auc_score(y_test_all, y_prob_imp)

    improved_results.append({
        'Model': name,
        'Threshold': round(best_thresh, 3),
        'Precision': round(precision, 4),
        'Recall': round(recall, 4),
        'F1 Score': round(f1, 4),
        'ROC AUC': round(auc, 4)
    })

    print(f"\n{'='*50}")
    print(f"{name} (optimal threshold={best_thresh:.3f}):")
    print(confusion_matrix(y_test_all, y_pred_imp))
    print(classification_report(y_test_all, y_pred_imp))
    print(f"ROC AUC: {auc:.4f}")

improved_df = pd.DataFrame(improved_results).sort_values('F1 Score', ascending=False)
print("\n" + "="*60)
print("IMPROVED MODEL COMPARISON (with threshold tuning):")
print("="*60)
print(improved_df.to_string(index=False))
```

## Isolation Forest — Unsupervised Anomaly Detection

```python
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_recall_curve

# Isolation Forest works on the ORIGINAL data (no SMOTE needed — it's unsupervised)
# Using the same scaled train/test split from the improved section above

iso_forest = IsolationForest(
    n_estimators=300,
    max_samples='auto',
    contamination=0.00173,  # matches the actual fraud ratio in dataset (~0.173%)
    random_state=42,
    n_jobs=-1
)

# Fit on training data only
iso_forest.fit(X_train_scaled)

# Predict: Isolation Forest returns -1 for anomalies, 1 for normal
iso_preds_raw = iso_forest.predict(X_test_scaled)
# Convert: -1 (anomaly) → 1 (fraud), 1 (normal) → 0 (legit)
iso_preds = np.where(iso_preds_raw == -1, 1, 0)

# Anomaly scores (more negative = more anomalous)
iso_scores = -iso_forest.decision_function(X_test_scaled)  # negate so higher = more fraudulent

print("Isolation Forest Results:")
print("="*50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, iso_preds))
print("\nClassification Report:")
print(classification_report(y_test_all, iso_preds))
print(f"ROC AUC Score: {roc_auc_score(y_test_all, iso_scores):.4f}")

# Threshold tuning on anomaly scores for better F1
prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, iso_scores)
f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
best_idx = np.argmax(f1_arr)
best_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.0

iso_preds_tuned = (iso_scores >= best_thresh).astype(int)
print(f"\nAfter Threshold Tuning (threshold={best_thresh:.4f}):")
print("="*50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, iso_preds_tuned))
print("\nClassification Report:")
print(classification_report(y_test_all, iso_preds_tuned))

report = classification_report(y_test_all, iso_preds_tuned, output_dict=True)
print(f"Recall: {report['1']['recall']:.4f}  |  F1: {report['1']['f1-score']:.4f}  |  Precision: {report['1']['precision']:.4f}")
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_recall_curve

# Convert to PyTorch tensors
X_train_tensor = torch.FloatTensor(X_train_res)
y_train_tensor = torch.FloatTensor(y_train_res.values if hasattr(y_train_res, 'values') else y_train_res)
X_test_tensor = torch.FloatTensor(X_test_scaled)

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_loader = DataLoader(train_dataset, batch_size=256, shuffle=True)

# MLP Architecture: 128 → 64 → 32 → 1
class FraudMLP(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.BatchNorm1d(128),
            nn.Dropout(0.3),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.BatchNorm1d(64),
            nn.Dropout(0.3),

            nn.Linear(64, 32),
            nn.ReLU(),
            nn.BatchNorm1d(32),
            nn.Dropout(0.2),

            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x).squeeze()

model_mlp = FraudMLP(X_train_scaled.shape[1])
print(model_mlp)

# Class weight for loss function
n_neg = np.sum(y_train_res == 0)
n_pos = np.sum(y_train_res == 1)
pos_weight = torch.FloatTensor([n_neg / n_pos])
criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
optimizer = torch.optim.Adam(model_mlp.parameters(), lr=0.001)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=5, factor=0.5)

# Training loop with early stopping
best_loss = float('inf')
patience, patience_counter = 10, 0
train_losses = []
val_losses = []

# Split a validation set from training data
val_size = int(0.15 * len(X_train_tensor))
indices = torch.randperm(len(X_train_tensor))
val_idx, train_idx = indices[:val_size], indices[val_size:]

X_val_t = X_train_tensor[val_idx]
y_val_t = y_train_tensor[val_idx]
train_sub = TensorDataset(X_train_tensor[train_idx], y_train_tensor[train_idx])
train_loader = DataLoader(train_sub, batch_size=256, shuffle=True)

for epoch in range(100):
    model_mlp.train()
    epoch_loss = 0
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        output = model_mlp(X_batch)
        loss = criterion(output, y_batch)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_train_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_train_loss)

    # Validation
    model_mlp.eval()
    with torch.no_grad():
        val_output = model_mlp(X_val_t)
        val_loss = criterion(val_output, y_val_t).item()
    val_losses.append(val_loss)
    scheduler.step(val_loss)

    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}/100 — Train Loss: {avg_train_loss:.4f} | Val Loss: {val_loss:.4f}")

    # Early stopping
    if val_loss < best_loss:
        best_loss = val_loss
        patience_counter = 0
        best_state = model_mlp.state_dict().copy()
    else:
        patience_counter += 1
        if patience_counter >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break

model_mlp.load_state_dict(best_state)
print("Training complete!")
```

```python
# Plot training history
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(train_losses, label='Train Loss')
ax.plot(val_losses, label='Val Loss')
ax.set_title('MLP Training History')
ax.set_xlabel('Epoch')
ax.set_ylabel('Loss')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# MLP Evaluation with threshold tuning
model_mlp.eval()
with torch.no_grad():
    mlp_probs = model_mlp(X_test_tensor).numpy()

# Default threshold (0.5)
mlp_preds_default = (mlp_probs >= 0.5).astype(int)
print("MLP Neural Network — Default Threshold (0.5):")
print("="*50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, mlp_preds_default))
print("\nClassification Report:")
print(classification_report(y_test_all, mlp_preds_default))
print(f"ROC AUC Score: {roc_auc_score(y_test_all, mlp_probs):.4f}")

# Optimal threshold tuning
prec_arr, rec_arr, thresholds = precision_recall_curve(y_test_all, mlp_probs)
f1_arr = 2 * prec_arr * rec_arr / (prec_arr + rec_arr + 1e-10)
best_idx = np.argmax(f1_arr)
mlp_best_thresh = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

mlp_preds_tuned = (mlp_probs >= mlp_best_thresh).astype(int)
print(f"\nMLP Neural Network — Optimized Threshold ({mlp_best_thresh:.4f}):")
print("="*50)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_all, mlp_preds_tuned))
print("\nClassification Report:")
print(classification_report(y_test_all, mlp_preds_tuned))

mlp_report = classification_report(y_test_all, mlp_preds_tuned, output_dict=True)
print(f"Recall: {mlp_report['1']['recall']:.4f}  |  F1: {mlp_report['1']['f1-score']:.4f}  |  Precision: {mlp_report['1']['precision']:.4f}")
```
